```python
# Imports
import pretty_midi
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Load one MIDI file - change the path to match your folder
midi_path = r"D:\IIT\4th Year\FYP\DATASETS\maestro-v3.0.0-midi\maestro-v3.0.0\2018\MIDI-Unprocessed_Chamber3_MID--AUDIO_10_R3_2018_wav--1.midi"
```

```python
# Reading the notes and plotting dynamics + tempo
midi = pretty_midi.PrettyMIDI(midi_path)

# Extract note velocities (dynamics)
velocities = []
times = []

for instrument in midi.instruments:
    for note in instrument.notes:
        velocities.append(note.velocity)
        times.append(note.start)

# Sort by time
sorted_pairs = sorted(zip(times, velocities))
times, velocities = zip(*sorted_pairs)

# Plot dynamics over time
plt.figure(figsize=(12, 4))
plt.plot(times, velocities)
plt.xlabel("Time (seconds)")
plt.ylabel("Velocity (dynamics)")
plt.title("Dynamics over time")
plt.show()

# Extract tempo
tempo_change_times, tempos = midi.get_tempo_changes()
print("Tempo changes:")
for t, bpm in zip(tempo_change_times, tempos):
    print(f"  Time: {t:.2f}s | BPM: {bpm:.1f}")
```
